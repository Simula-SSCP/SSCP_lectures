# Exercise 15.1a solution

**Left-hand side:**

$$p \cdot C_{\mathrm{m}} \frac{\partial V}{\partial t} \rightarrow \text{cm} \cdot \frac{\mu\text{F}}{\text{cm}^2} \cdot \frac{\text{mV}}{\text{ms}} = \frac{\mu\text{A}}{\text{cm}}$$

$$p \cdot I_{\mathrm{ion}} \rightarrow \text{cm} \cdot \frac{\mu\text{A}}{\text{cm}^2} = \frac{\mu\text{A}}{\text{cm}}$$

**Right-hand side:**

$$\frac{1}{r_i + r_e} \frac{\partial^2 V}{\partial x^2} \rightarrow \frac{1}{\text{k}\Omega/\text{cm}} \cdot \frac{\text{mV}}{\text{cm}^2} = \frac{\text{cm}}{\text{k}\Omega} \cdot \frac{\text{mV}}{\text{cm}^2} = \frac{\text{mV}}{\text{k}\Omega \cdot \text{cm}} = \frac{\mu\text{A}}{\text{cm}}$$

Both sides have units of $\mu\text{A}/\text{cm}$, which can be interpreted as current per unit length along the cable. ✓
