# Exercise 8.2f solution

Playing around with the widget, we find that we must max out the potassium conductance (e.g., $g_{\mathrm{K}} \approx 25\text{–}30$ µS) and minimize the calcium and sodium conductances (e.g., $g_{\mathrm{Na}} = 1$ µS and $g_{\mathrm{Ca}} = 0$ µS) to end up at a resting potential of -80 mV.

This shouldn't be too surprising, as -80 mV is very close to the Nernst potential for potassium (-86 mV), meaning potassium must be absolutely dominating the tug-of-war when the cell is at rest.

In real biological practice, we find that the potassium conductance at rest is indeed about 25 times that of the combined sodium and calcium conductances. This is because specific "leak" potassium channels are almost always open at rest, while calcium and sodium channels tend to be tightly closed.
