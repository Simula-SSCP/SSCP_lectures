# Exercise 8.1f solution


```python
# --- Setup code from previous sub-exercises ---
import numpy as np

F = 96485  # C/mol
R = 8.314  # J/(mol K)
T = 310  # K
# ----------------------------------------------
```

```python
# Normal physiological conditions
E_Na = R * T / F * np.log(110 / 8) * 1e3
print(f"Original E_Na:              {E_Na:.1f} mV")

# 30% increase in intracellular Na (8 mM * 1.3 = 10.4 mM)
E_Na_new = R * T / F * np.log(110 / (8 * 1.3)) * 1e3
print(f"After 30% increase in Na:   {E_Na_new:.1f} mV")

print(f"Difference:                 {E_Na_new - E_Na:.1f} mV")
```
