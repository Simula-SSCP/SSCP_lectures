# Exercise 8.2a solution

We will solve the following ODE numerically:

$$
C_{\mathrm{m}}\frac{\mathrm{d}V}{\mathrm{d}t} = - g_{\mathrm{Na}} (V - E_{\mathrm{Na}})
$$


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp


# Define the ODE
def rhs(t, V, Cm, g_Na, E_Na):
    dV_dt = -g_Na * (V - E_Na) / Cm
    return dV_dt


# Define the parameters and initial condition
T = (0, 100)
Cm = 0.05  # nF
E_Na = 70.0  # mV (Copied from computation in previous exercise)
g_Na = 0.005  # µS
params = (Cm, g_Na, E_Na)
y0 = (-80,)  # mV

# Solve the ODE
solution = solve_ivp(rhs, T, y0, args=params, max_step=0.1)

t = solution.t
(V,) = solution.y

plt.plot(t, V)
plt.axhline(E_Na, linestyle="--", color="black", label=r"$E_{\mathrm{Na}}$")
plt.xlabel("Time (ms)")
plt.ylabel("Membrane Potential (mV)")
plt.legend()
plt.show()
```
