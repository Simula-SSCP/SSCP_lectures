# Exercise 8.1b solution

Looking at the concentrations of the three ions, we see that there is a massive inward concentration gradient for sodium and calcium. This is why their Nernst potentials are positive: the cell requires a highly positive internal electrical potential to physically repel the positive ions and balance out the inward concentration gradient.

For potassium, however, the concentration gradient points in the opposite direction (it wants to leave the cell). Therefore, we need an electrical potential difference in the opposite direction (a negative internal potential) to pull the positively charged potassium back in and balance the gradient.
