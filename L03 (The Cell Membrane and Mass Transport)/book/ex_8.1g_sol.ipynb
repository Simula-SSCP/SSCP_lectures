# Exercise 8.1g solution

Based on our calculations, we saw in **(8.1e)** that a full action potential barely changes the intracellular concentration at all ($\sim$ 0.01% change). Furthermore, we saw in **(8.1f)** that even if we _force_ a massive 30% change in the intracellular concentration, the Nernst potential only shifts by a few millivolts due to the logarithmic nature of the Nernst equation.

Because the massive fluxes of ions during an action potential are still incredibly small relative to the bulk volume of the cell, the concentrations remain practically undisturbed. Therefore, during normal short-term physiological activity, it is a very safe and highly accurate assumption to treat the Nernst potentials as strict mathematical constants.
