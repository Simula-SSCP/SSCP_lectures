# Exercise 8.1d solution

We now find the number of individual ions 6.0 pC corresponds to. Using Faraday's constant:


```python
# --- Setup code from previous sub-exercises ---
Cm = 0.05  # nF
dV = 120  # mV
dQ = Cm * dV
# ----------------------------------------------
```

```python
F = 96485  # C/mol

# Convert charge to moles (pC / (C/mol) = pmol)
dNa = dQ / F
print(f"Roughly {dNa:.2e} picomoles pass through the membrane.")
```

Again, this number ($\sim 6 \times 10^{-5}$ picomoles) _seems_ super small, but we need to view it in the context of the cell's volume to really understand its impact.
