# Exercise 8.2e solution


From the plot, we can see that the equilibrium looks to be around -65 mV. If we want to write it out exactly, we can read the last value of the `V` array. Because it is a NumPy array, using negative indexing (`[-1]`) lets us read the very last element:


```python
# --- Setup code from previous sub-exercises ---
import numpy as np
from scipy.integrate import solve_ivp


def rhs(t, V, Cm, g_Na, g_K, g_Ca, E_Na, E_K, E_Ca):
    return -(g_Na * (V - E_Na) + g_K * (V - E_K) + g_Ca * (V - E_Ca)) / Cm


T = (0, 50)
Cm = 0.05
g_Na = 0.005
g_K = 0.05
g_Ca = 0.002
E_Na = 70.0
E_K = -86.0
E_Ca = 114.0
params = (Cm, g_Na, g_K, g_Ca, E_Na, E_K, E_Ca)
y0 = (0,)

solution = solve_ivp(rhs, T, y0, args=params, max_step=0.1)
(V,) = solution.y
# ----------------------------------------------
```

```python
print(f"Equilibrium potential: {float(V[-1]):.1f} mV")
```

Rerunning the code with a different initial condition shows us that the final equilibrium potential **does not change**. The initial condition only changes where the curve starts, not where it ends up.

The equilibrium potential becomes -65 mV in this case because this is a balancing point between the three competing currents. Each current individually wants the membrane potential to be equal to its own specific Nernst potential. The sodium and calcium currents are pulling $V$ upwards (positive), while the potassium current is pulling it down (negative).

Because we have set the potassium conductance to be so much stronger than the others ($g_{\mathrm{K}} \gg g_{\mathrm{Na}} \approx g_{\mathrm{Ca}}$), it easily wins this "tug-of-war" and the equilibrium lies close to the Nernst potential for potassium.
