# Exercise 8.3: A super simple action potential model

So far we have modeled how the membrane potential reaches its equilibrium potential as a smooth exponential relaxation. However, in practice, what we really want to model are _action potentials_, which are spontaneous, rapid spikes in the membrane potential. These action potentials occur because different ion channels in the membrane physically open and close over time, dynamically changing the equilibrium potential.

To properly understand action potentials, we will need to understand exactly when and why ion channels open or close, which is called ion channel _gating_. This is the topic for the next lecture! For now, however, let us simply state that sodium channels in the membrane open for a short, fixed duration, and see how this mathematically affects the membrane potential.

Let us model a membrane with sodium and potassium channels (no calcium channels this time) so our ODE is:

$$
C_{\mathrm{m}}\frac{\mathrm{d}V}{\mathrm{d}t} = -g_{\mathrm{Na}} (V - E_{\mathrm{Na}}) - g_{\mathrm{K}} (V - E_{\mathrm{K}})
$$

Now we assume that $g_{\mathrm{K}}$ is constant, meaning the potassium channels are always open and not actively gated. The sodium conductance, however, is time-dependent and equal to:

$$
g_{\mathrm{Na}}(t) = \begin{cases}
0.55 \text{ µS} & \text{if } 2 \text{ ms} < t < 2.2 \text{ ms}, \\
0.003 \text{ µS} & \text{otherwise.}
\end{cases}
$$

This corresponds to the membrane having a massive number of sodium channels that are mostly closed, but that collectively open for a very brief duration (0.2 ms), temporarily skyrocketing the sodium conductance.

Here we have created a very simple phenomenological model of an action potential by stating that, at a specific time, sodium channels pop open and stay open for a fixed, short duration. While this is a gross simplification of how sodium channels actually work, it produces an action potential curve that is remarkably correct.
