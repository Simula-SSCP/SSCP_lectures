# Exercise 7.2b solution

$$
\frac{\mathrm{d}V}{\mathrm{d}t} = \frac{V_{\mathrm{target}} - V}{C_{\mathrm{m}} R_{\mathrm{s}}}
$$

We start by looking at the numerator. Because we are computing $V_{\mathrm{target}} - V$, these two variables must be in the exact same units. We want to use millivolts here (mV).

For the denominator, we compute the product of $C_{\mathrm{m}}$, which is in nanofarads (nF), and $R_{\mathrm{s}}$, which is in megaohms (M$\Omega$). The product $C_{\mathrm{m}}R_{\mathrm{s}}$ elegantly evaluates to milliseconds (ms). _(If you find these unit computations tricky, you can always use a tool like Wolfram Alpha to double-check!)_

The right-hand side (RHS) is now in units of **mV/ms**. If we let our time $t$ also be given in milliseconds, this perfectly matches our choice of measuring the membrane potential rate of change in mV/ms on the left-hand side.

Therefore, the equation is naturally consistent, and no mathematical changes or scaling factors need to be applied.
