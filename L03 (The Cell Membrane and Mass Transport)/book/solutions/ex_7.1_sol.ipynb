# Exercise 7.1 solution

The major thing to get right here is to be strictly consistent with the units. The precise computation itself is not overly important; we are just looking for a ballpark estimate of the ratios.


```python
for case, radius, width in (
    ("Apple", 35, 150e-3),
    ("Earth", 6400, 25),
    ("Cell", 10, 5e-3),
):
    print(f"Relative width of {case}'s exterior: {width / radius:.3%}")
```

We see that the three exteriors are all extremely thin compared to their respective radii, but especially so for the cell membrane, which is a full order of magnitude relatively thinner than an apple's skin or the Earth's crust!
