# Exercise 8.1e solution


```python
# --- Setup code from previous sub-exercises ---
Cm = 0.05  # nF
dV = 120  # mV
dQ = Cm * dV

F = 96485  # C/mol

# Convert charge to moles (pC / (C/mol) = pmol)
dNa = dQ / F
# ----------------------------------------------
```

```python
V_cell = 20  # picoliters (pL)

# pmol / pL mathematically evaluates directly to mmol/L (mM)
dconc = dNa / V_cell * 1e3
rel = 1 - (8 - dconc) / 8  # Relative change in intracellular sodium

print(f"Absolute concentration change: {dconc:.4f} mM")
print(f"Relative concentration change: {rel:.3%}")
```

Now we have arrived at a relative measure! We see the change in concentration is on the order of a hundredth of a percent (0.01%), which is an absolutely negligible change in the cell's bulk concentration.
