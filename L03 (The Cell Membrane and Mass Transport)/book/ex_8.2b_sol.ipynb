# Exercise 8.2b solution

The membrane potential stabilizes at 70 mV, which is our exact value for $E_{\mathrm{Na}}$. For our chosen parameters, we see that it takes roughly 25–50 ms to reach this value. Because the mathematical shape is an exponential relaxation, it is technically impossible to set an exact discrete time point where the potential has fully "equilibrated," but it is practically there after 50 ms.
