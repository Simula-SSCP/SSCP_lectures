# Exercise 20.1: Thermodynamics of calcium pumps

In this exercise, you will analyze the thermodynamic limits of the SERCA pump and the NCX exchanger using pen and paper (or Markdown!).


## Exercise 20.1a: SERCA and Gibbs Energy

For every cycle, the SERCA pump moves two calcium ions from the cytosol into the SR. This incurs a change in Gibbs energy of:

$$
\Delta G_{\mathrm{pump}} = 2RT \ln\left(\frac{[\mathrm{Ca}^{2+}]_{\mathrm{SR}}}{[\mathrm{Ca}^{2+}]_{\mathrm{i}}}\right)
$$

1. It is generally true that $[\mathrm{Ca}^{2+}]_{\mathrm{SR}} > [\mathrm{Ca}^{2+}]_{\mathrm{i}}$. What does this mathematically mean for the sign of $\Delta G_{\mathrm{pump}}$?
2. Recall that spontaneous processes require $\Delta G < 0$. What does your answer to #1 tell us about SERCA?


## Exercise 20.1b: The ATP limit

SERCA breaks down one ATP molecule to power this process. Therefore, the pump can only run as long as the total energy is negative:

$$
\Delta G_{\mathrm{pump}} + \Delta G_{\mathrm{ATP}} < 0
$$

By combining this with the equation from 20.1a, we can solve for the maximum SR concentration:

$$
[\mathrm{Ca}^{2+}]_{\mathrm{SR}} < [\mathrm{Ca}^{2+}]_{\mathrm{i}} \cdot \exp\left[\frac{-\Delta G_{\mathrm{ATP}}}{2RT}\right]
$$

Inside a mammalian myocyte, the energy released by ATP ($\Delta G_{\mathrm{ATP}}$) is roughly **-50 kJ/mol**. At body temperature, $2RT \approx 5.2$ kJ/mol.

Calculate the absolute maximum concentration ratio ($[\mathrm{Ca}^{2+}]_{\mathrm{SR}} / [\mathrm{Ca}^{2+}]_{\mathrm{i}}$) that SERCA can theoretically maintain.

_(A commonly cited experimental finding is that SERCA maintains a gradient of about 1:7000. How does the biological reality compare to your theoretical thermodynamic limit?)_


## Exercise 20.1c: Thermodynamics of the NCX

The NCX transports 3 $\mathrm{Na}^+$ ions inward and 1 $\mathrm{Ca}^{2+}$ ion outward. Because it relies entirely on electrochemical gradients (no ATP), the net change in Gibbs energy must equal 0 at equilibrium:

$$
\Delta G_{\mathrm{NCX}} = \Delta G_{\mathrm{Ca}} - 3\Delta G_{\mathrm{Na}} = 0
$$

The energy required to move an ion across the membrane is:

$$
\Delta G_{\mathrm{X}} = z_{\mathrm{X}} F (E_{\mathrm{X}} - V)
$$

Substitute the $\Delta G$ formulas for sodium ($z=1$) and calcium ($z=2$) into the NCX equation. Solve the algebra to find the reversal potential ($V = E_{\mathrm{NCX}}$) where the exchanger stops!
