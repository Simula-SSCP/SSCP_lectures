# Exercise 20.2f solution

Since the intracellular calcium concentration changes drastically during an action potential, so does $E_{\mathrm{Ca}}$ and hence $E_{\mathrm{NCX}}$.

During the upstroke/plateau of the AP, when $V$ is high and $[\mathrm{Ca}^{2+}]_{\mathrm{i}}$ is elevated, we can briefly have $V > E_{\mathrm{NCX}}$, meaning the NCX lets some calcium **enter** the cell (reverse mode).

During repolarisation, $V$ drops below $E_{\mathrm{NCX}}$ and the NCX switches to forward mode, pumping calcium **out** of the cell. This is its main role.
