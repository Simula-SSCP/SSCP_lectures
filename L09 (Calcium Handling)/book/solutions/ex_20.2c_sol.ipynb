# Exercise 20.2c solution

Inserting $\Delta G_{\mathrm{Na}} = z_{\mathrm{Na}}F(E_{\mathrm{Na}} - V)$ with $z_{\mathrm{Na}}=1$ and $\Delta G_{\mathrm{Ca}} = z_{\mathrm{Ca}}F(E_{\mathrm{Ca}} - V)$ with $z_{\mathrm{Ca}}=2$:

$$\Delta G_{\mathrm{NCX}} = 2F(E_{\mathrm{Ca}} - V) - 3F(E_{\mathrm{Na}} - V)$$

$$= F\big[2E_{\mathrm{Ca}} - 2V - 3E_{\mathrm{Na}} + 3V\big]$$

$$= F\big[V + 2E_{\mathrm{Ca}} - 3E_{\mathrm{Na}}\big]$$
