# Exercise 19.1i solution

By playing with the widget, we observe:

- If $K_d$ is very high, the CICR mechanism requires too much calcium to activate, and oscillations disappear.
- If $K_d$ is very low, CICR activates immediately and the system reaches a new (higher) steady state without oscillating.
- There is a 'sweet spot' for $K_d$ where oscillations appear, roughly around 0.3–0.8 µM.
- Increasing $n$ sharpens the sigmoidal curve, making the transition between low and high release more switch-like, which tends to promote oscillations.
