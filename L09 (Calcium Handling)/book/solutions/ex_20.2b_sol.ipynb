# Exercise 20.2b solution

For one transported calcium ion, three sodium ions are moved in the opposite direction. The Gibbs energy of the full cycle is the sum of the individual contributions:

$$\Delta G_{\mathrm{NCX}} = \Delta G_{\mathrm{Ca}} - 3\Delta G_{\mathrm{Na}}$$

The sign reflects the fact that sodium moves in the opposite direction to calcium.
