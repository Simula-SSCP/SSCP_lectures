```python
import math

# Luo-Rudy NCX implementation
# Given: V, Nai, Cai, Cao, Nao, k_NaCa, K_mNa, K_mCa, k_sat, eta, R, T, F

num = (Nai**3 * Cao) * math.exp(eta * V * F / (R * T)) - (Nao**3 * Cai) * math.exp(
    (eta - 1) * V * F / (R * T)
)

den = (
    (K_mNa**3 + Nao**3)
    * (K_mCa + Cao)
    * (1 + k_sat * math.exp((eta - 1) * V * F / (R * T)))
)

i_NaCa = k_NaCa * num / den
```
