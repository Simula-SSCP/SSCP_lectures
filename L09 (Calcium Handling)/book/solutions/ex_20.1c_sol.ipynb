Starting from $\Delta G_{\mathrm{pump}} + \Delta G_{\mathrm{ATP}} < 0$:

$$2RT \ln\left(\frac{[\mathrm{Ca}^{2+}]_{\mathrm{SR}}}{[\mathrm{Ca}^{2+}]_{\mathrm{i}}}\right) + \Delta G_{\mathrm{ATP}} < 0$$

$$\ln\left(\frac{[\mathrm{Ca}^{2+}]_{\mathrm{SR}}}{[\mathrm{Ca}^{2+}]_{\mathrm{i}}}\right) < \frac{-\Delta G_{\mathrm{ATP}}}{2RT}$$

$$\frac{[\mathrm{Ca}^{2+}]_{\mathrm{SR}}}{[\mathrm{Ca}^{2+}]_{\mathrm{i}}} < \exp\left(\frac{-\Delta G_{\mathrm{ATP}}}{2RT}\right)$$

Inserting $\Delta G_{\mathrm{ATP}} \approx -50$ kJ/mol and $2RT \approx 5.2$ kJ/mol:

```python
import numpy as np

dG_ATP = -50  # kJ/mol
two_RT = 5.2  # kJ/mol

max_ratio = np.exp(-dG_ATP / two_RT)
print(f"Maximum concentration ratio: {max_ratio:.0f}")
```

The exact answer is highly sensitive to the value of $\Delta G_{\mathrm{ATP}}$ because it enters exponentially. With 50 kJ/mol we get ~15000; with 45 kJ/mol we get ~5700. The order of magnitude (thousands) is what matters.
