# Exercise 19.1c solution

The second term is the **Hill equation**. It starts near zero for low $[\mathrm{Ca}^{2+}]_{\mathrm{i}}$ and increases sigmoidally, saturating at $\kappa_1$ for high calcium. The total $k_{\mathrm{rel}} = \kappa_0 + \kappa_1 \cdot \text{(Hill term)}$ represents a baseline leak ($\kappa_0$) plus a calcium-dependent CICR component.
