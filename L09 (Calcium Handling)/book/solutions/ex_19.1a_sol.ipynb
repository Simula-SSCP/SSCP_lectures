# Exercise 19.1a solution

1. Since $[\mathrm{Ca}^{2+}]_{\mathrm{o}} > [\mathrm{Ca}^{2+}]_{\mathrm{i}}$, the difference is positive, so $J_{\mathrm{entry}} > 0$ (positive when calcium enters the cytosol).

2. $J_{\mathrm{rel}} = k_{\mathrm{rel}} \left([\mathrm{Ca}^{2+}]_{\mathrm{SR}} - [\mathrm{Ca}^{2+}]_{\mathrm{i}}\right)$

3. $J_{\mathrm{extrusion}} = k_{\mathrm{extrusion}} \cdot [\mathrm{Ca}^{2+}]_{\mathrm{i}}$,  $J_{\mathrm{uptake}} = k_{\mathrm{uptake}} \cdot [\mathrm{Ca}^{2+}]_{\mathrm{i}}$
