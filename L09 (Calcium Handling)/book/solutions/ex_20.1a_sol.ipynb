# Exercise 20.1a solution

Since $[\mathrm{Ca}^{2+}]_{\mathrm{SR}} > [\mathrm{Ca}^{2+}]_{\mathrm{i}}$, the fraction inside the logarithm is greater than 1, and $\ln(\cdot) > 0$.

Therefore $\Delta G_{\mathrm{pump}} > 0$. This means the process does not occur spontaneously — SERCA must be an example of **active transport**, not passive. Moving calcium against its concentration gradient requires an external energy source.
