# Exercise 21.1d solution

The total transmembrane current is:

$$I_{\mathrm{m}} = i_{\mathrm{Na}} + i_{\mathrm{Ca,L,Ca}} + i_{\mathrm{Ca,L,K}} + i_{\mathrm{K}} + i_{\mathrm{NaCa}} + i_{\mathrm{K1}} + i_{\mathrm{Kp}} + i_{\mathrm{p,Ca}} + i_{\mathrm{Na,b}} + i_{\mathrm{Ca,b}} + i_{\mathrm{NaK}} + i_{\mathrm{ns,Na}} + i_{\mathrm{ns,K}}$$

And the voltage ODE:

$$\frac{dV}{dt} = \frac{I_{\mathrm{stim}} - I_{\mathrm{m}}}{C_m}$$
