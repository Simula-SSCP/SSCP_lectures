# Exercise 20.2d solution

Setting $\Delta G_{\mathrm{NCX}} = 0$:

$$2F(E_{\mathrm{Ca}} - V) = 3F(E_{\mathrm{Na}} - V)$$

$$2E_{\mathrm{Ca}} - 2V = 3E_{\mathrm{Na}} - 3V$$

$$V = 3E_{\mathrm{Na}} - 2E_{\mathrm{Ca}}$$

This is the reversal potential of the NCX: $E_{\mathrm{NCX}} = 3E_{\mathrm{Na}} - 2E_{\mathrm{Ca}}$.

When $V < E_{\mathrm{NCX}}$, the NCX operates in forward mode (Ca$^{2+}$ out).
When $V > E_{\mathrm{NCX}}$, it operates in reverse mode (Ca$^{2+}$ in).
