# Exercise 21.1c solution

The four calcium compartment ODEs:

$$\frac{d[\mathrm{Ca}^{2+}]_{\mathrm{SS}}}{dt} = B_{\mathrm{SS}} \left( \frac{J_{\mathrm{rel}} V_{\mathrm{JSR}}}{V_{\mathrm{SS}}} - \frac{J_{\mathrm{xfer}} V_{\mathrm{myo}}}{V_{\mathrm{SS}}} - i_{\mathrm{Ca,L,Ca}} \cdot \text{conv\_Amp\_SS} \right)$$

$$\frac{d[\mathrm{Ca}^{2+}]_{\mathrm{JSR}}}{dt} = B_{\mathrm{JSR}} \left( J_{\mathrm{tr}} - J_{\mathrm{rel}} \right)$$

$$\frac{d[\mathrm{Ca}^{2+}]_{\mathrm{NSR}}}{dt} = \frac{(J_{\mathrm{up}} - J_{\mathrm{leak}}) V_{\mathrm{myo}}}{V_{\mathrm{NSR}}} - \frac{J_{\mathrm{tr}} V_{\mathrm{JSR}}}{V_{\mathrm{NSR}}}$$

$$\frac{d[\mathrm{Ca}^{2+}]_{\mathrm{i}}}{dt} = B_i \left( J_{\mathrm{leak}} + J_{\mathrm{xfer}} - J_{\mathrm{up}} - J_{\mathrm{trpn}} - (i_{\mathrm{Ca,b}} - i_{\mathrm{NaCa}} + i_{\mathrm{p,Ca}}) \cdot \text{conv\_Amp\_myo} \right)$$
