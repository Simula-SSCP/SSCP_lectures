# Exercise 20.2a solution

The NCX as a whole is an example of **active transport**, because it moves calcium against its concentration gradient. However, it achieves this by coupling to the **passive** transport of sodium (down its concentration gradient). So one can say NCX combines passive sodium transport with active calcium transport.

In forward mode: passive Na$^+$ influx drives active Ca$^{2+}$ efflux.  
In reverse mode: passive Ca$^{2+}$ influx drives active Na$^+$ efflux.
