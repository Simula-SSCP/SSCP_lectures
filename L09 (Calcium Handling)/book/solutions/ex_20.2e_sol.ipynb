```python
E_Na = 70  # mV
E_Ca = 114  # mV

E_NCX = 3 * E_Na - 2 * E_Ca
print(f"E_NCX = {E_NCX} mV")
```

With these values, $E_{\mathrm{NCX}} \approx -18$ mV. Since the resting membrane potential (~-80 mV) is well below this, the NCX is predominantly in forward mode at rest, pumping calcium out.
