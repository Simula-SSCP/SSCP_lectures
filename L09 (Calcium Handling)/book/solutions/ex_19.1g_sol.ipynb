# Exercise 19.1g solution

```python
import numpy as np
import matplotlib.pyplot as plt

kappa0 = 0.013
kappa1 = 0.58
Kd = 0.5
n = 3.0

Cai = np.linspace(0, 2, 1000)
k_rel = kappa0 + kappa1 * Cai**n / (Kd**n + Cai**n)

plt.plot(Cai, k_rel)
plt.xlabel("[Ca2+]i (µM)")
plt.ylabel("k_rel (1/s)")
plt.xlim(0, 2)
plt.show()
```

Caffeine makes the CICR steeper. This corresponds to increasing $\kappa_1$ (stronger maximum CICR) or decreasing $K_d$ (lower calcium threshold for CICR).
