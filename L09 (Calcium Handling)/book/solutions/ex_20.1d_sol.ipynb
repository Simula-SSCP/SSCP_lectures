# Exercise 20.1d solution

The commonly cited experimental gradient is about 1:7000. Our theoretical limit of ~15000 is in the right ballpark. The fact that the measured gradient is somewhat below the theoretical limit indicates that SERCA is operating close to — but not at — its thermodynamic limit, which makes sense since no molecular machine is perfectly efficient.

This exercise demonstrates (1) using Gibbs energy to make practical estimates, and (2) that SERCA maintains enormous concentration gradients.
