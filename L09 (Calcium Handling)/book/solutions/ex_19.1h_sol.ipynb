# Exercise 19.1h solution

By exploring the widget:

- **$\kappa_0$** should be **low** — this is the baseline leak, and we want minimal release at low calcium.
- **$\kappa_1$** should be **high** — this sets the maximum CICR-driven release rate.
- **$K_d$** should be **moderate** — it sets the calcium concentration at which half-maximal release occurs. If it is too low, the system fires at very low calcium; if too high, the CICR never activates strongly.
- **$n$** should be **high** — increasing the Hill coefficient makes the transition from low to high release steeper, giving a more switch-like behavior.
