# Exercise 21.1a solution

The RyR 4-state Markov model ODEs:

$$\frac{dP_{C1}}{dt} = -k_a^+ \cdot [\mathrm{Ca}^{2+}]_{\mathrm{SS}}^n \cdot P_{C1} + k_a^- \cdot P_{O1}$$

$$\frac{dP_{O1}}{dt} = k_a^+ \cdot [\mathrm{Ca}^{2+}]_{\mathrm{SS}}^n \cdot P_{C1} - k_a^- \cdot P_{O1} - k_b^+ \cdot [\mathrm{Ca}^{2+}]_{\mathrm{SS}}^m \cdot P_{O1} + k_b^- \cdot P_{O2} - k_c^+ \cdot P_{O1} + k_c^- \cdot P_{C2}$$

$$\frac{dP_{O2}}{dt} = k_b^+ \cdot [\mathrm{Ca}^{2+}]_{\mathrm{SS}}^m \cdot P_{O1} - k_b^- \cdot P_{O2}$$

$$\frac{dP_{C2}}{dt} = k_c^+ \cdot P_{O1} - k_c^- \cdot P_{C2}$$

The **subspace** calcium ($\mathrm{Ca}^{2+}_{\mathrm{SS}}$) determines the opening, because the RyR channels are located at the dyadic cleft.

Release flux: $J_{\mathrm{rel}} = v_1 \cdot (P_{O1} + P_{O2}) \cdot ([\mathrm{Ca}^{2+}]_{\mathrm{JSR}} - [\mathrm{Ca}^{2+}]_{\mathrm{SS}})$
