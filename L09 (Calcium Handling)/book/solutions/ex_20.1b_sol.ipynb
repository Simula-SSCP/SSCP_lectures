# Exercise 20.1b solution

SERCA is an ATPase: it couples the thermodynamically unfavourable pumping to the hydrolysis of ATP. Each individual cycle must still satisfy the laws of thermodynamics, so the total change in Gibbs energy must be negative:

$$\Delta G_{\mathrm{pump}} + \Delta G_{\mathrm{ATP}} < 0$$

In other words, SERCA cannot spend more energy pumping calcium than it harvests from ATP.
