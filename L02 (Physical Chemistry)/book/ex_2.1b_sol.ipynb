```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp


def rhs(t, y, kp, km):
    # Unpack state vector
    MbO2, Mb, O2 = y

    # Compute derivatives
    dMbO2_dt = kp * Mb * O2 - km * MbO2
    dMb_dt = -dMbO2_dt
    dO2_dt = -dMbO2_dt

    # Return RHS as sequence
    return (dMbO2_dt, dMb_dt, dO2_dt)
```
