# Exercise 5.1g solution


**Discussion:**

The reaction velocity starts high because there is a lot of substrate available to convert. As time progresses, the total amount of enzyme remains constant, but the amount of free substrate decreases, resulting in a lower overall reaction velocity. The fact that the reaction velocity drops right from the start indicates that the enzyme is not fully saturated; otherwise, the reaction velocity would have stayed nearly constant at $v_{\rm max}$ during the beginning phase.

The two curves are slightly different at the very start of the simulation. This is because our numerical simulation is based strictly on the law of mass action, whereas the Michaelis-Menten equation assumes the system is in a quasistatic steady state (i.e., the enzyme-substrate complex is always at equilibrium).

We can see that this quasistatic assumption is poor right at the start of the reaction because our initial conditions are far from equilibrium. However, the system very _quickly_ reaches this steady state, and from that point on, the simplified Michaelis-Menten equation serves as an excellent approximation.
