# Exercise 2.1d solution


```python
# --- Setup code from previous sub-exercises ---
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp


def rhs(t, y, kp, km):
    # Unpack state vector
    MbO2, Mb, O2 = y

    # Compute derivatives
    dMbO2_dt = kp * Mb * O2 - km * MbO2
    dMb_dt = -dMbO2_dt
    dO2_dt = -dMbO2_dt

    # Return RHS as sequence
    return (dMbO2_dt, dMb_dt, dO2_dt)


# Define time span
time = (0, 5)

# Define initial condition vector
y0 = (0, 2, 4)

# Define parameter vector
kp = 2
km = 1
params = (kp, km)

# Call the ODE solver
solution = solve_ivp(rhs, time, y0, args=params, max_step=0.1)
# ----------------------------------------------
```

```python
# Split up the solution matrix
t = solution.t
MbO2, Mb, O2 = solution.y

# Plot the solutions
plt.plot(t, MbO2, label="MbO2")
plt.plot(t, Mb, label="Mb")
plt.plot(t, O2, label="O2")

# Prettify the figure
plt.xlabel("Time (arb. units)")
plt.ylabel("Concentration (arb. units)")
plt.legend()

# Show the figure
plt.show()
```
