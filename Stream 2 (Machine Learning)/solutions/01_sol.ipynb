# Python classes — Solutions

```python
import numpy as np
import matplotlib.pyplot as plt
```

## Exercise 1: A `LinearModel` class

```python
class LinearModel:
    def __init__(self, slope, intercept):
        self.slope = slope
        self.intercept = intercept

    def predict(self, x):
        return self.slope * x + self.intercept
```

```python
model = LinearModel(slope=2.5, intercept=-1.0)

# Test on a few values
for x in [0, 1, 2, 3]:
    print(f"predict({x}) = {model.predict(x)}")

# Plot
x = np.linspace(0, 5, 100)
plt.plot(x, model.predict(x))
plt.xlabel("x")
plt.ylabel("y")
plt.title("LinearModel: y = 2.5x - 1.0")
plt.show()
```

## Exercise 2: Fitting the model to data

```python
# Generate noisy linear data
rng = np.random.default_rng(42)
x_data = np.linspace(0, 5, 50)
y_data = 3.0 * x_data + 1.0 + rng.normal(0, 1.5, size=len(x_data))
```

```python
class LinearModel:
    def __init__(self, slope=0.0, intercept=0.0):
        self.slope = slope
        self.intercept = intercept

    def predict(self, x):
        return self.slope * x + self.intercept

    def fit(self, x, y):
        x_mean = np.mean(x)
        y_mean = np.mean(y)
        self.slope = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean) ** 2)
        self.intercept = y_mean - self.slope * x_mean
```

```python
model = LinearModel()
model.fit(x_data, y_data)
print(f"Fitted slope: {model.slope:.2f} (true: 3.0)")
print(f"Fitted intercept: {model.intercept:.2f} (true: 1.0)")

x_line = np.linspace(0, 5, 100)
plt.scatter(x_data, y_data, alpha=0.5, label="Data")
plt.plot(x_line, model.predict(x_line), color="red", label="Fitted line")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.title("OLS fit")
plt.show()
```

## Exercise 3: Dunder methods

```python
class LinearModel:
    def __init__(self, slope=0.0, intercept=0.0):
        self.slope = slope
        self.intercept = intercept

    def predict(self, x):
        return self.slope * x + self.intercept

    def fit(self, x, y):
        x_mean = np.mean(x)
        y_mean = np.mean(y)
        self.slope = np.sum((x - x_mean) * (y - y_mean)) / np.sum((x - x_mean) ** 2)
        self.intercept = y_mean - self.slope * x_mean

    def __repr__(self):
        return f"LinearModel(slope={self.slope:.2f}, intercept={self.intercept:.2f})"

    def __call__(self, x):
        return self.predict(x)
```

```python
model = LinearModel(slope=2.5, intercept=-1.0)

# __repr__: just type the variable name
print(model)

# __call__: use the object as a function
print(model(3.0))  # same as model.predict(3.0)
print(model.predict(3.0))
```
