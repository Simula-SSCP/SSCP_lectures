## Exercise 3: Special methods ("dunder" methods)

Python classes can define **special methods** (also called *dunder* methods because they start and end with double underscores) that let your objects interact naturally with Python's built-in syntax. For example:

- `__repr__(self)` defines what happens when you type the object's name in a cell or call `print(model)`. It should return a string.
- `__call__(self, ...)` defines what happens when you use the object as a function, e.g., `model(x)`.
- `__add__(self, other)` defines what happens when you add the object with another object, e.g., `a + b`.
- `__mul__(self, other)` defines what happens when you multiply the object with another object, e.g., `c * d`.
