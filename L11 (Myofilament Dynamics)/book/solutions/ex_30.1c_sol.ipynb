# Exercise 30.1c solution


1. Could this model reproduce the steep, sigmoidal force-pCa curve? What mechanisms would need to be added?

No. The simple model uses constant kinetic rates and thus produces a shallow, Michaelis-Menten-like response to calcium. To match the experimentally observed steep, switch-like activation, we need cooperativity — where the activation of one RU or XB makes it easier for its neighbours to activate too.

2. Why is cooperativity important?

Cooperativity provides positive feedback: as more sites activate, it becomes easier for remaining sites to turn on. This transforms a gradual response into a steep, all-or-nothing switch, which is essential for efficient cardiac contraction. Without cooperativity, the Hill coefficient of the force-pCa curve would be ~1, far below the experimentally observed values of 3–7.
