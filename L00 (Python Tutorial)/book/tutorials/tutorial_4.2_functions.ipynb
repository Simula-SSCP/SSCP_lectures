# Defining functions

So far, we have used many built-in functions, for example `print`, `type`, `np.linspace`, and `plt.plot`. All of these are tools we can _call on_ to perform specific tasks. Now we will look at how we can define our _own_ functions.

The benefit of defining our own functions is that we can avoid writing the same code over and over again. This makes our code much more efficient and easier to fix if something goes wrong. It also makes our code vastly more structured and readable because a function provides a layer of **abstraction**. Once we write a function that performs a specific task, we can rely on that function later without having to think about the messy underlying mathematical details.

Let us show an example using our trusty kinematics formula:

$$
s(t) = v_0t + \frac{1}{2}at^2
$$

To define this formula as a reusable Python function, we write:


````python
v0 = 12
a = -9.81


def s(t):
    return v0 * t + 0.5 * a * t**2
````

Note the following syntax rules:

- We define a function using the keyword `def`.
- The name of the function is followed by parentheses and a colon `:`.
- The inputs to the function are represented by _arguments_ written inside the parentheses.
- The entire contents of the function must be indented.
- We define the final output of the function with the keyword `return`.

In this case, `s` is the _name_ of the function. Inside the function, we compute our formula as normal, but we use the keyword `return` to send the computed value back to whatever called it.

When you run the code cell above, absolutely nothing appears to happen. This is because defining a function does exactly that—it simply _defines_ the rules in memory. To actually use it, we must **call** the function:


````python
s(0.4)
````

Here, we pass the value `0.4` to our function. Python computes the defined formula using $t=0.4$ and returns the result.

The true benefit is that we can now use this clean `s(t)` function in our loops without cluttering our code with the actual math:


````python
dt = 0.1

for i in range(10):
    t = i * dt
    print(f"{t:3.1f}  {s(t):3.1f}")
````

## Function arguments and defaults

In our previous example, we let `v0` and `a` be defined _outside_ the function itself. These are called **global variables**. While relying on global variables is practical for quick, messy scripts, it is considered bad practice for larger programs because it can cause unpredictable behavior.

A better solution is to let the parameters be passed as explicit arguments to the function:

```python
def s(t, v0, a):
    return v0 * t + 0.5 * a * t**2
```

However, this can get annoying because we can no longer simply call `s(t)`. We would be forced to pass all three parameters every single time we call the function. The ultimate solution is to assign **default values** to the arguments:


````python
def s(t, v0=12, a=-9.81):
    return v0 * t + 0.5 * a * t**2
````

Here, we give `v0` and `a` default fallback values. It is now entirely optional whether we supply them when calling the function.


````python
print(s(0.4))  # Uses the default v0 and a
print(s(0.4, v0=5.0, a=0.0))  # Overrides the defaults
````

Most of the built-in functions we have used so far utilize optional arguments. For instance, the `sorted()` function sorts a list in increasing order by default. However, we can use an optional argument called `reverse` to flip its behavior:


````python
print(sorted([4, 0, 1, 3, 5, 2]))
print(sorted([4, 0, 1, 3, 5, 2], reverse=True))
````

## A more complex example: Checking if a number is prime

Functions can contain _any_ logic, including loops and if-statements; they do not need to be strict mathematical formulas. Let's write a function that checks if a number is a prime number.

A prime number is any integer larger than 1 that is only cleanly divisible by 1 and itself. First, we need to handle the special case that 1 and below are not prime:

```python
def is_prime(n):
    if n <= 1:
        return False
```

For numbers larger than 1, we need to see if any number in the range $[2, n)$ divides into it evenly. We do this by looping through all candidates `d` and checking if the remainder `n % d` is exactly zero:


````python
def is_prime(n):
    # Rule out 1, 0, and negative numbers
    if n <= 1:
        return False

    # Loop through all potential divisors
    for d in range(2, n):
        if n % d == 0:
            # If it divides cleanly, it is NOT prime!
            return False

    # If the loop finishes and nothing divided cleanly, it MUST be prime!
    return True
````

A few details to note here:

- The final line `return True` is un-indented back to the primary function level, meaning it is _not_ part of the for-loop. It will only execute if the entire loop finishes without finding a divisor.
- As soon as a function hits a `return` statement, **it immediately terminates**. So as soon as we find a candidate `d` that divides `n`, the function returns `False` and abruptly stops. It doesn't waste time checking the rest of the numbers.

Let us test our newly created function using a loop!


````python
for n in range(1, 11):
    if is_prime(n):
        print(f"{n} is prime!")
    else:
        print(f"{n} is not prime.")
````
