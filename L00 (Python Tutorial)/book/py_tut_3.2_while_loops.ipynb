# While loops

We now turn to the topic of loops. Loops are one of the most fundamental concepts of programming, providing a powerful way to repeat operations rapidly. In Python, we have two main kinds of loops: _while-loops_ and _for-loops_. We start by looking at while-loops.

A while-loop repeats itself as long as a specific condition remains fulfilled. After each iteration of the loop, the condition is checked again. If the condition is still true, the loop is run again. If the condition is no longer true, the loop terminates.

The syntax of a while-loop in Python is as follows:

```python
while <condition>:
    <code block to be repeated>
    <code block to be repeated>
```

There are a few critical things to note with this syntax:

- We define the start of the loop with the keyword `while`.
- We write in a given condition, and end the line with a colon (`:`).
- A set of _indented statements_ follows the definition. These are the lines that execute each time the loop iterates.

The lines with indentation belong to the while-loop. Any statements without indentation will only be executed _after_ the loop has finished running completely.

## Example: Computing $s(t)$ over time

Returning to our recurring kinematics example:
$$s(t) = v_0 t + \frac{1}{2}at^2$$

So far, we have only computed specific single values for a given $t$. But what if we want to print out a whole table of values representing an object's trajectory over time?

Let us change the scenario so we are throwing a ball straight up in the air ($a = -9.81$, $v_0 = 12$). We want to compute $s(t)$ over the first two seconds of the ball's flight. Let us increment our time by $\Delta t = 0.1$ seconds for each step.


````python
# Parameters
v0 = 12
a = -9.81

# Time parameters
dt = 0.1  # Increment per time step
t = 0  # Starting time
T = 2  # Final time

# Loop
while t <= T:
    s = v0 * t + 0.5 * a * t**2
    print(f"{t:2.1f}  {s:4.2f}")
    t = t + dt
````

In this loop, the time variable `t` starts at 0, and the loop continues as long as the condition `t <= T` evaluates to true. During each iteration, we increment `t` by `dt`. This means every time the loop runs, `t` grows by 0.1 until it eventually breaches `T` and the loop ends.

In addition to incrementing `t`, the loop computes $s(t)$ and prints both variables in a neatly formatted table. The format specifiers `:2.1f` and `:4.2f` force the text to maintain a specific width, keeping the columns perfectly aligned.

### Is there an error in our code?

There is something slightly strange about our output. According to the while-loop condition, we should keep going while $t \leq T$ (where $T=2$). This means the final pass of the loop should include `t = 2.0`. However, in the printed table, we see that it only goes up to $t = 1.9$ before stopping. Why did this happen?

To understand this, we must remember that computers cannot represent decimal numbers flawlessly in binary. Therefore, the increments of `0.1` have not been represented with infinite precision. Consequently, by the time the loop reaches the 2-second mark, the internal variable `t` is actually ever so slightly _larger_ than 2.0, causing the `t <= T` condition to fail!

We can prove this by printing `t` to a machine precision of 16 decimal places:


````python
print(f"{t:.16f}")
````

Because of this inherent floating-point limitation, you must always be very careful when testing or comparing exact decimals in `while` conditions!

## Example: Computing until the ball hits the ground

In our previous example, we ran the loop for exactly two seconds. But at $t = 2$, the ball had not yet hit the ground. What if we want the code to keep computing _until_ the ball hits the ground?

This is exactly what while-loops are best at! We just change the condition to `while s >= 0`. As soon as the height $s$ becomes negative, the loop stops. We do need to define an initial dummy value for `s` before the loop starts so the computer has something to check the first time it encounters the condition.


````python
# Parameters
v0 = 12
a = -9.81
dt = 0.1
t = 0

s = 0  # Initial dummy value so the loop can start

while s >= 0:
    s = v0 * t + 0.5 * a * t**2
    print(f"{t:2.1f}  {s:4.2f}")
    t = t + dt
````

We can clearly see that $t=2.5$ is the first time step where the ball has dropped below $s=0$. This illustrates a major strength of while-loops: they allow us to compute toward a specific physical condition even if we have no idea how many steps it will take to get there!
