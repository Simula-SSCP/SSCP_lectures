# Lists and sequences

Earlier, we mentioned that Python has built-in data types for collections of values. These collections are incredibly useful for storing datasets that naturally belong together, such as a time series of measurements.

One of the most widely used collections in Python is the **list** (often generally referred to as a sequence). Lists are simply ordered sequences of values. We can define a list by placing values within square brackets `[]` and separating them with commas.


```python
primes = [2, 3, 5, 8, 11, 13, 17]
```

Now, `primes` is a list variable. We say that it consists of 7 _elements_.

We can find the total length of a list using the built-in `len()` function. We can also access individual elements inside the list as if they were independent variables through a process called **indexing**. We do this by writing the name of the list followed by square brackets containing the index number, for example: `primes[2]`.

**Important:** Python starts counting at 0! Therefore, `[2]` will extract the _third_ element in the list, not the second.


```python
print(type(primes))  # Check the data type of the variable
print(len(primes))  # Print the number of elements
print(primes[2])  # Access an element by index
```

## Appending elements

Python lists have a series of **methods**—special functions directly attached to the object that allow us to alter it. The most common method is `.append()`, which we can use to add a new element to the end of a list. We can start with an empty list and dynamically build it up element by element using a loop!

Let us say we want to generate a list of all powers of 2 that are less than 1000 ($2^n < 1000$). We can use a while-loop and append the elements one by one:


```python
powers = []  # Start with an empty list
n = 0

while 2**n < 1000:
    powers.append(2**n)
    n += 1

# Print the final populated list
print(powers)
```

## Other useful list methods

As shown above, we can use `.append(value)` to add elements. Because lists are highly versatile, they possess several other useful methods:

| Method           | Description                                             |
| ---------------- | ------------------------------------------------------- |
| `.append(value)` | Add a new element to the end of the list.               |
| `.sort()`        | Sort the list numerically or alphabetically _in-place_. |
| `.reverse()`     | Reverse the current order of the list _in-place_.       |
| `.pop()`         | Removes the _last_ element of the list and returns it.  |
| `.pop(index)`    | Removes the element at a specific index and returns it. |
| `.remove(value)` | Removes the _first_ occurrence of a specific value.     |

For a more thorough description of Python lists and their methods, you can refer to the [official Python Tutorial](https://docs.python.org/3/tutorial/introduction.html#lists).
