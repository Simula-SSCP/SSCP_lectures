# A quick introduction to Jupyter notebooks

In Jupyter, files are known as _notebooks_, and are saved on your computer as `.ipynb` files. These are not traditional scripts or programs, but rather a combination of rich text, math, figures, and code all contained in the same file. You interact with this file through your web browser, which enables Jupyter to use HTML and other useful formatting features.

In this school, you will run on a cloud-based Jupyter server, meaning the actual Python code isn't run by your machine. If you install and run Jupyter locally on your machine, it will look exactly the same, and you still use your browser to interact with the notebook, but this time the Jupyter server is your own machine.

## Notebook cells

The contents of a notebook are divided into a series of _cells_. A cell is a collection of content, and they are split into two main types. **Markdown cells** contain content we associate with a typical text, such as written text, math, figures, tables, and so on. **Code cells** contain Python code which can be executed, and the output is written directly into the notebook right below the code itself.

## Execution order of cells

One element of notebooks that can be a bit confusing, to beginners and experts alike, is that a notebook is one big interactive session. This means that variables and results are remembered _after_ running a cell and can be used further down in a notebook in different code cells.

However, when shutting down the notebook altogether and reloading it, this interactive data is forgotten. When opening a notebook, one can often _not_ run a single cell in the middle or towards the end of the notebook; rather, one should run all the cells in the notebook from the start. A helpful tool here is to use the `Cell > Run All` option from the toolbar, which will run all cells in order and load everything into memory. Alternatively, you can use `Kernel > Restart & Run All` if you want to completely reset and then run all cells in order.

If you start a new session in an existing notebook and execute cells _out of order_, you might get bugs or weird results. The typical example is simply error messages saying variables don't exist as they have not been defined or imported (because an earlier cell that does this has not been executed). In weird cases, the code might run, but the variables might not be what was originally intended, and so the results are off.

Therefore: **Remember to execute existing cells _in order_.**
