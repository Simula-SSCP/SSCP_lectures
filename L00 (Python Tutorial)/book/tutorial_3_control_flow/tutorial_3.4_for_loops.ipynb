# For loops

We have covered while-loops; now we turn to **for-loops**. While a while-loop repeats blindly until a condition is met, a for-loop systematically goes through a predefined collection of objects, such as a list. We say that we _iterate_ through a list.

The syntax for a for-loop in Python is as follows:

```python
for <element> in <sequence>:
    <codeblock to be repeated>
```

Here, `<sequence>` should be some object consisting of multiple items (like a list). The `<element>` is a temporary variable name that we choose. This variable is automatically assigned to each item in the list sequentially as the loop repeats.

We can, for example, easily loop through and print all the elements of our `primes` list from the previous section:


````python
primes = [2, 3, 5, 8, 11, 13, 17]

for p in primes:
    print(p)
````

The variable `p` did _not_ exist before we created the loop; the loop creates it. Think of `p` as a temporary mathematical variable that updates its value for every iteration. Reading the code out loud as an English command makes perfect sense: _"For each element `p` in the sequence `primes`, print `p`."_

Let us perform a slightly more mathematical computation inside the loop. We can explore Mersenne primes, which take the form $2^p - 1$ where $p$ is a prime number:


````python
print(" p      2^p - 1")
print("---------------")

for p in primes:
    print(f"{p:2} {2**p - 1:10}")
````

## Looping over number ranges

Often, we do not want to loop over a manually predefined list of elements, but rather over a massive, continuous range of numbers.

One way to accomplish this would be to manually type out a massive list (`[0, 1, 2, 3...]`), but this is incredibly inefficient. Luckily, Python supplies a highly optimized `range()` function that creates sequential number ranges on the fly.

If we want to loop over the integers 1 through 10, we do:


````python
for i in range(1, 11):
    print(i)
````

Notice that `range(1, 11)` starts at 1, but ends at **10**. The two arguments dictate the start (inclusive) and the end (exclusive). Mathematically, this evaluates the interval $[1, 11)$.

Also, note that we called our iterating variable `i`. This is standard programming convention when counting upward in integer steps, originating from mathematical summation notation:
$$\sum_{i=0}^{N} 2i - 1$$

To compute this exact sum in Python, we would simply define an accumulator variable and use a for-loop:


````python
N = 100
total = 0

for i in range(0, N + 1):
    total += 2 * i - 1

print(total)
````

We used `range(0, N+1)` because we wanted to start at 0 and include exactly $N$. However, whenever we want a range to start at `0`, we can simply omit the first argument as a handy shorthand! `range(N+1)` and `range(0, N+1)` are identical in Python.

We can also add a third argument to dictate the _step size_: `range(start, stop, step)`. If we only want even numbers, we could step by 2:


````python
for i in range(0, 10, 2):
    print(i)
````

## Using a for-loop to compute $s(t)$

Earlier, we used a while-loop to compute our kinematics formula $s(t)$ for discrete time steps. Let us repeat this exercise using a for-loop.

The main challenge is that `range()` only generates _integers_, but we want time $t$ to increase in decimal steps of $0.1$. To get around this, we can let our loop variable `i` count in integer steps, and then manually compute $t_i = i \cdot \Delta t$ inside the loop!

This time, instead of just printing the values, let's heavily utilize list methods and append the generated values into two lists so we can save the data for later.


````python
# Parameters
v0 = 12
a = -9.81
dt = 0.1
N = 21

# Empty lists to store our data
t_list = []
s_list = []

for i in range(N):
    t = i * dt
    s = v0 * t + 0.5 * a * t**2

    t_list.append(t)
    s_list.append(s)
````

We have now stored the results inside `t_list` and `s_list`. In the next chapter, we will use these exact lists to plot a graph.

But for now, if we want to print out our neat table using lists, we have a problem: we have two separate lists. We can loop over both of them simultaneously by _zipping_ them together using Python's highly useful `zip()` function:


````python
for t, s in zip(t_list, s_list):
    print(f"{t:3.1f}   {s:3.1f}")
````
