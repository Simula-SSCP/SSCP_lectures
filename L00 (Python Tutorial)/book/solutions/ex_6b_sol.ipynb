```python
# --- Setup code from previous sub-exercises ---
import numpy as np


def is_prime(n):
    # We only need to check factors up to the square root of n.
    # If a number has no factors up to its square root, it is prime.
    if n <= 1:
        return False

    max_factor = int(np.sqrt(n) + 1)

    for d in range(2, max_factor):
        if n % d == 0:
            return False

    return True


# ----------------------------------------------
```

```python
prime_counter = 0
number = 2

while prime_counter < 10001:
    if is_prime(number):
        prime_counter += 1
    number += 1

# Since 'number' is incremented at the end of the loop,
# the actual prime is 'number - 1'
final_prime = number - 1
print(f"The 10001st prime is {final_prime}.")
```
