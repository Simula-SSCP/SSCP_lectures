```python
# --- Setup code from previous sub-exercises ---
from math import pi
# ----------------------------------------------
```

```python
# First convert radius to meters
R_m = 6371 * 1000  # m

# Then compute volume in m^3
V_m3 = (4 / 3) * pi * R_m**3

# Finally convert to liters
V_liters = V_m3 * 1000  # 1 cubic meter = 1000 liters

print(f"The volume of Earth is {V_liters:.3e} liters")
```
