```python
# --- Setup code from previous sub-exercises ---
import numpy as np
import matplotlib.pyplot as plt


def f(x):
    return x**2 + 4 * x - 6


def g(x):
    return -2 * x**2 + 6 * x + 6


x = np.linspace(-5, 5, 10000)
# ----------------------------------------------
```

```python
plt.plot(x, f(x))
plt.plot(x, g(x))
plt.show()
```
