```python
# --- Setup code from previous sub-exercises ---
from math import pi

R_m = 6371 * 1000
V_m3 = (4 / 3) * pi * R_m**3
V_liters = V_m3 * 1000
# ----------------------------------------------
```

```python
M = 5.972e24  # kg
density = M / V_liters

print(f"The average density of Earth is {density:.3f} kg/L")
```
