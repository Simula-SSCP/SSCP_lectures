# Variables and data types

Let us return to our recurring kinematics example:
$$s(t) = v_0 t + \frac{1}{2}at^2$$

Earlier, we performed this computation by defining four variables:


```python
t = 0.5
v0 = 2
a = 0.2
s = v0 * t + 0.5 * a * t**2
```

Here, we defined `t`, `v0`, `a`, and `s`. Notice that we never had to declare the _type_ of the variables before making them. Python handles this automatically dynamically! This is very different from strictly typed programming languages like Java or C++, where you must explicitly declare if a variable is an integer or a decimal before using it.

Even though we do not declare the types manually, every variable still possesses a specific _data type_ under the hood. You can check the type of any variable using the built-in `type()` function.


```python
print(type(t))
print(type(v0))
print(type(a))
print(type(s))
```

We can see that `v0` became an `int` (an integer variable), while the other three variables became `float`s. A _float_ is a decimal number. Because computers cannot perfectly represent arbitrary decimal numbers with infinite precision in binary, they are called "floating-point numbers" as they have microscopic inaccuracies.

There are several built-in data types in Python. The most common fundamental ones are:

| Name           | Python type | Example values    |
| -------------- | ----------- | ----------------- |
| Boolean        | `bool`      | `True` / `False`  |
| Integer        | `int`       | `4`               |
| Floating-point | `float`     | `3.1415`          |
| Complex number | `complex`   | `2.5 + 4j`        |
| Text String    | `str`       | `"Hello, World!"` |

These are the foundational data types, and they each consist of one specific, single value. In addition, there are many built-in data types designed for storing _collections_ of values. Some of these are:

| Name       | Python Type | Example                   |
| ---------- | ----------- | ------------------------- |
| Tuple      | `tuple`     | `(0, 0, 1)`               |
| List       | `list`      | `[2, -3, 5, 8]`           |
| Dictionary | `dict`      | `{"a": -9.81, "v0": 5.2}` |
| Set        | `set`       | `{2, 3, 5, 8, 11}`        |

These container structures behave differently from one another, and we will explore them further in the next chapter.

When you use external libraries, they usually introduce their own highly optimized data types. The most important one for this course is from the `numpy` (Numerical Python) package, which introduces the `ndarray` data type. These represent mathematical vectors, matrices, and tensors, and are extremely useful for performing fast linear algebra operations. We will use `numpy` arrays extensively in this course!


## Print formatting

So far, we have seen how we can define variables, perform simple computations, and look at the raw results using `print`. However, we usually want more control over how our output is formatted so that it is readable by humans. We accomplish this using **print formatting**.

We already know we can print variables directly:


```python
print(t)
print(s)
```

While this works, it lacks context. We can write a more descriptive message by injecting variables directly into a text string using **f-strings** (formatted strings).


```python
print(f"After {t} seconds, the object has traveled {s} meters.")
```

Notice that we placed an `f` immediately _before_ the opening quotation mark of the string. This tells Python to look for curly braces `{}` inside the text and evaluate whatever variables or code are placed inside them.

When using f-strings, you can precisely control the formatting of each variable by adding a colon `:` followed by a format specification inside the braces:


```python
print(f"After {t:.2f} seconds, the object has traveled {s:.1f} meters.")
```

Here, we specified that we want the time written out with exactly two decimals by writing `{t:.2f}`. The `f` tells Python to format it as a _float_, and the `.2` specifies two decimal places. Similarly, we print out `s` with a single decimal.

When printing with a restricted number of decimals, Python automatically rounds the number for us.


```python
x = 2.1436
print(f"{x:.3f}")
print(f"{x:.2f}")
```

Notice that when rounding to three decimals, Python rounds _up_, while for two decimals, it rounds _down_.

Python uses a convention known as [bankers' rounding](https://en.wikipedia.org/wiki/Rounding), in which a `.5` is rounded to the nearest _even_ number. This is the standard, statistically unbiased way to round numbers for large data sets.

You can also write numbers out in scientific notation automatically:


```python
n = 41828
print(f"{n:.1e}")
```

The format specifier `.1e` states that we want the number formatted in scientific notation with a single decimal.

Another highly convenient specifier is `%`, which automatically multiplies a decimal number by 100 and adds the percentage sign! No need to manually multiply it yourself. For example, if we have performed 913 trials and 642 were successful:


```python
success = 642
total = 913
ratio = success / total

print(f"Success rate: {ratio:.1%}")
```
