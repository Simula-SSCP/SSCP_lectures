# If-tests

An if-test allows us to endow our programs with logic, allowing the code to check if a specific condition is met and only carry out tasks if that condition is true.

The normal syntax of a simple if-test in Python is:

```python
if <condition>:
    <do something>
```

Here, `<condition>` is anything that can be evaluated to either `True` or `False`. If the condition is true, the code inside the test runs (typically called the _if-block_). If the condition is false, the code is skipped and not executed.

You can identify which code belongs to a given conditional because it is given an **indentation**. This means the code lines are moved slightly inward from the left-hand margin. This is accomplished by using the `Tab` key on the keyboard. (Most editors, including Jupyter, will automatically add this indentation for you when you press Enter after writing an `if` statement).

The indented code block belongs to the condition and will execute if and only if the condition is true. Any code lines further down _without_ indentation no longer belong to the test and will execute no matter what.

Let us look at an example. Suppose we have a dataset of $n$ samples, and we want to find the median. If the number of samples is an odd number, the median is simply the middle value. But if it is an even number of samples, we need to compute the average of the two middle values. To program an algorithm for this, we must _test_ if $n$ is even or odd.


````python
n = 23

if n % 2 == 0:
    print(f"{n} is even")
else:
    print(f"{n} is odd")
````

In this example, we first define `n = 23`. Then we write a test to check if it is an even number using `if n % 2 == 0`. This might look cryptic, so let us break it down:

- `n % 2` means "$n$ modulo 2," which gives the remainder when $n$ is divided by 2.
- An even number is defined as any integer perfectly divisible by 2, meaning the remainder is exactly 0.
- To check if two numbers are exactly equal, we use double equality signs (`==`). We must do this because a single equality sign (`=`) is reserved for assigning variables!

When the if-test is encountered in the code, the condition is checked. Because `n` isn't even, the condition evaluates to `False`. We can check this ourselves:


````python
print(n % 2 == 0)
````

Because the condition is `False`, the code indented under the `if` statement is completely ignored. However, we also added an `else` block. The `else` does not have its own condition; it simply acts as a catch-all that executes if the initial condition was false.

_(Note: You do not always need to include an `else`. If there is no `else` and the condition is false, the code simply does nothing and moves on)._

## Other comparisons

In our example, we checked if two numbers were equal with `==`. You can also check if two numbers are _not equal_ using `!=`. Furthermore, you can check which is larger or smaller using `<` and `>`. For greater-than-or-equal, use `>=`.

You can also combine several conditions using the `and` and `or` operators. The following test, for example, checks for a number that is both _positive_ and _even_:


````python
n = 10

if n > 0 and n % 2 == 0:
    print(f"{n} is both positive and even")
````

## More than two outcomes

We have shown that you can use `if-else` to have two distinct outcomes. However, sometimes we need to handle multiple distinct possibilities. In Python, we can add secondary conditions to a test by using the keyword `elif` (short for "else if").

Say, for instance, we need to know if a number is positive, negative, or exactly zero:


````python
x = -2

if x > 0:
    print(f"{x} is greater than 0")
elif x == 0:
    print(f"{x} is exactly 0")
elif x < 0:
    print(f"{x} is smaller than zero")
````

A given test block can only have a single `if` and a single `else`, but it can have as many `elif`s as you want. However, only a _single_ block will ever be executed! If the first `if` clause is true, all the subsequent `elif` blocks will be entirely skipped, even if their conditions also happen to be true.

If you want different conditions to be checked completely independently, you simply use multiple separate `if` statements in a row.
