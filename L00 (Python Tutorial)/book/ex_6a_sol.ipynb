The first solution is based on the lecture notes, while the second solution is an optimized alternative.


```python
def is_prime(n):
    # Check if divisible for all smaller numbers (based on lecture notes)
    if n <= 1:
        return False
    for d in range(2, n):
        if n % d == 0:
            return False

    return True
```

```python
import numpy as np


def is_prime(n):
    # Optimization: We only need to check factors up to the square root of n.
    # If a number has no factors up to its square root, it is prime.
    if n <= 1:
        return False

    max_factor = int(np.sqrt(n) + 1)

    for d in range(2, max_factor):
        if n % d == 0:
            return False

    return True
```
