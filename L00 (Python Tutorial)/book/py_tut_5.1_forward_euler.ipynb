# Solving an ODE numerically

Lastly, we want to show an example of how we can solve an ordinary differential equation (ODE) numerically using Python. This is a massive mathematical topic, and we definitely won't do it full justice here, but this serves as a practical example of advanced mathematical computation.

As a starting example, we want to solve a simple _exponential decay_ problem:

$$
\frac{\mathrm{d}u}{\mathrm{d}t} = -au
$$

where $a$ is a constant.

To solve such an ODE numerically, it is common to discretize it so that we solve it for specific, discrete time points:

$$
t_i = i \cdot \Delta t
$$

for some small time step $\Delta t$. We then define $u_i$ as the numerical solution at time $t_i$, meaning:

$$
u_i \approx u(t_i)
$$

To step the solution forward in time, we approximate the exact derivative with the _Forward Euler_ finite difference:

$$
\frac{u_{i+1} - u_{i}}{\Delta t} = -a u_i
$$

Solving this algebraically for the future time step $u_{i+1}$ gives:

$$
u_{i+1} = u_{i} - a u_i \cdot \Delta t
$$

Because we know the solution at the current time $t_i$, this formula allows us to compute the solution one step forward into the future. By placing this formula inside a loop, we can chain these steps together and compute the trajectory across time!

To solve an actual case, we need to define the parameter $a$, the time step $\Delta t$, and we must choose an initial starting condition $u_0$.


```python
import numpy as np
import matplotlib.pyplot as plt

# Define time step and number of steps
dt = 0.01  # time step
T = 10  # end time
n = int(T / dt)  # total number of steps

# Define parameters and initial condition
a = 0.4
u0 = 4

# Create empty numpy arrays to store the solutions
t = np.zeros(n + 1)
u = np.zeros(n + 1)

# Set the initial condition at index 0
u[0] = u0

# Use a loop to solve the equation one step at a time
for i in range(n):
    t[i + 1] = t[i] + dt
    u[i + 1] = u[i] - a * u[i] * dt

# Plot the computed solution
plt.plot(t, u)
plt.xlabel("Time (t)")
plt.ylabel("State (u)")
plt.title("Forward Euler: Exponential Decay")
plt.show()
```
