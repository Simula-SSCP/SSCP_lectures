# Plotting

We now turn to a highly rewarding topic: Plotting. Plotting is the process of producing visual graphics from numerical data. Generating plots is one of the most popular use cases for Python, and it is absolutely essential for scientific programming and data science.

Basic Python does not have built-in plotting capabilities, so we need to import external functionality. Different packages for plotting exist, but we will use the most commonly used one: [Matplotlib](https://matplotlib.org/).

Matplotlib is a powerful, open-source package for 2D plotting. To work with it, we first need to import its `pyplot` subpackage. It is standard convention to import it under the shortened alias `plt`:


```python
import matplotlib.pyplot as plt
```

We can now use `plt.` to access plotting commands.

Let's assume you still have the `t_list` and `s_list` variables generated from the for-loop in the previous chapter. The standard `plt.plot()` command plots two sequences against each other: the first along the x-axis, and the second along the y-axis. After configuring the plot, we must call `plt.show()` to actually render the figure to the screen.


```python
# Recreating the lists from the previous chapter just in case they were lost
v0 = 12
a = -9.81
dt = 0.1
N = 21

t_list = []
s_list = []

for i in range(N):
    t_list.append(i * dt)
    s_list.append(v0 * (i * dt) + 0.5 * a * (i * dt) ** 2)
```

```python
# Plotting the data
plt.plot(t_list, s_list)
plt.show()
```

This curve shows our data in an intuitive way. However, the axes are not labeled, which is a terrible practice in scientific modeling! Let's fix this by adding some additional commands between the `plot` and `show` functions to format the figure.


```python
plt.plot(t_list, s_list)
plt.xlabel("Time (s)")
plt.ylabel("Height (m)")
plt.title("Motion over time")
plt.show()
```

Even though our dataset consists of discrete points in time, the plotted curve appears continuous. This is because Matplotlib automatically draws a straight line between each consecutive data point for us. We can add a formatting string like `"o--"` to show the individual data points explicitly:


```python
plt.plot(t_list, s_list, "o--")
plt.xlabel("Time (s)")
plt.ylabel("Height (m)")
plt.title("Motion over time")
plt.show()
```

## Using NumPy for math and plotting

The package [NumPy](http://www.numpy.org/) (Numerical Python) is critical if we want to do any heavy numerical computation in Python. In our $s(t)$ example, we manually used a loop and lists to compute our data points. With NumPy, we can do this computation directly using vectors.

First, we import it (standard convention is to use the alias `np`):


```python
import numpy as np
```

Now we can create a NumPy _array_. These behave much like standard Python lists, but they are strictly mathematical in nature, meaning we can perform _vectorized computations_ on them! We simply write out the math, and NumPy handles the loops blazing-fast behind the scenes.

There are three main ways to define NumPy arrays:

- `np.zeros(n)`: Creates an array of $n$ zeros.
- `np.arange(start, stop, step)`: Works exactly like `range()`, but returns an array and accepts decimal steps.
- `np.linspace(a, b, n)`: Stands for "linear spacing". It generates exactly $n$ evenly spaced points in the closed interval $[a, b]$.

Using NumPy, we can compute our entire $s(t)$ trajectory in just a few lines of code without a single manual loop:


```python
v0 = 12
a = -9.81

# Create an array of times from 0 to 3.0 in steps of 0.1
t = np.arange(0, 3.1, 0.1)

# Compute the entire array of heights in one mathematical sweep!
s = v0 * t + 0.5 * a * t**2
```

Now, `t` is no longer a single number, but an _array_ (a mathematical vector). Because of this, the resulting variable `s` is automatically computed as an array of the exact same size:


```python
print(type(t))
print(type(s))
```

```python
plt.plot(t, s)
plt.xlabel("t")
plt.ylabel("s(t)")
plt.axis([0, 3.0, 0, 8.5])
plt.show()
```

NumPy also provides vectorized versions of all standard mathematical functions (`np.sin`, `np.exp`, `np.log`, `np.sqrt`). If we want to create a sine curve, we can do it instantly:


```python
x = np.linspace(0, 4 * np.pi, 50)
y = np.sin(x)

plt.plot(x, y)
plt.xlabel("x")
plt.ylabel("sin(x)")
plt.show()
```

Notice that we used `linspace` to plot two whole periods (from 0 to $4\pi$). The curve looks nice, except the peaks look a bit jagged. This is because 50 points are simply too few to make it look smooth. Let's up the resolution to 1001 points:


```python
x = np.linspace(0, 4 * np.pi, 1001)
y = np.sin(x)

plt.plot(x, y)
plt.show()
```

## Other kinds of plots

We have only shown the command `plt.plot` so far to plot line curves. However, Matplotlib supports nearly any kind of 2D plot imaginable. Some popular commands include:

- `plt.bar` for bar charts
- `plt.scatter` for scatter plots
- `plt.hist` for histograms
- `plt.contour` for contour plots
