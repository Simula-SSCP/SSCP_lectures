**1.** The voltage must cross the threshold (~−33 mV) for the Na⁺ positive feedback to drive it past 0 mV. Starting from rest, the effective time constant is $\tau \approx C_m/g_K = 0.5$ ms, so V nearly reaches steady state during the 1 ms pulse. **2.** The threshold is approximately $V_0 \approx -33$ mV, found analytically as the unstable root of $g_{\mathrm{Na}} \, m_\infty(V)(V - E_{\mathrm{Na}}) + g_K(V - E_K) = 0$. Below this voltage, the K⁺ leak dominates and V relaxes back to rest. Above it, Na⁺ positive feedback overwhelms the leak and drives V toward ~36 mV. In the $I(V)$ plot, the threshold corresponds to the middle zero crossing (positive slope = unstable equilibrium), flanked by two stable equilibria (negative slope) at rest and the excited state.

**3.** This model cannot generate a complete action potential because there is no mechanism to bring $V$ back down. Without inactivation ($h$) or a delayed K⁺ current ($n$), the Na⁺ gate stays open and $V$ saturates at the excited state near 36 mV.
