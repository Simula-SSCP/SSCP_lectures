Key observations:

**1. Rheobase:** The minimum stimulus current to trigger an AP depends on stimulus duration. For a 1 ms pulse, approximately $I_{\mathrm{app}} \approx 7$ µA/cm² is needed. For longer pulses, less current is required.

**2. Reducing $\bar{g}_{\mathrm{Na}}$:** As $\bar{g}_{\mathrm{Na}}$ decreases below ~40 mS/cm², the AP amplitude decreases. Below ~20 mS/cm², the cell fails to fire at all — the inward Na⁺ current is insufficient to overcome the outward K⁺ current.

**3. Increasing $\bar{g}_{\mathrm{K}}$:** Higher K⁺ conductance causes faster repolarization (shorter AP duration) and a deeper afterhyperpolarization. This is because the K⁺ current more forcefully pulls V toward $E_{\mathrm{K}}$.

**4. Repetitive firing:** With a sustained stimulus (long duration, e.g., 20 ms) and moderate current (e.g., $I_{\mathrm{app}} = 7-10$ µA/cm²), the cell can produce multiple APs. This occurs because after each AP, the $h$ gate recovers sufficiently for the sustained current to trigger another spike. The firing frequency increases with stimulus amplitude.
