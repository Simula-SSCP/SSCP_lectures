## 1. Temperature scaling

The temperature difference is $\Delta T = 37 - 6.3 = 30.7°\text{C}$, which is approximately 3.07 increments of $10°\text{C}$.

With $Q_{10} = 3$: total scaling factor $= 3^{3.07} \approx 28.5$.

All rate constants would be multiplied by $\approx 28.5$. Since the time constants are the inverse of the sum of rates, they would be divided by $\approx 28.5$. The action potential would be roughly 28x shorter — which is consistent with the observation that squid APs at $6.3°\text{C}$ last $\sim 2$ ms, while mammalian neuronal APs at $37°\text{C}$ last $\sim 0.1$ ms.

## 2. Faster $h$ gate

If $\tau_h$ is halved, inactivation would kick in earlier, cutting off the sodium current sooner. The upstroke would be smaller (less time for $\text{Na}^+$ to flow in before the channel inactivates), and the action potential peak would be lower. In pathological cases, an extremely fast $h$ gate could prevent firing altogether by shutting the channel before enough $\text{Na}^+$ current flows to sustain the positive feedback loop.
