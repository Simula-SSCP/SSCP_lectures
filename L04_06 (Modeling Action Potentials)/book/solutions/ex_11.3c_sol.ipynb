```python
# --- Setup code from previous sub-exercises ---
import numpy as np
import matplotlib.pyplot as plt


def alpha_m(V):
    return 0.1 * (V + 40) / (1 - np.exp(-(V + 40) / 10))


def beta_m(V):
    return 4.0 * np.exp(-(V + 65) / 18)


def alpha_h(V):
    return 0.07 * np.exp(-(V + 65) / 20)


def beta_h(V):
    return 1.0 / (1 + np.exp(-(V + 35) / 10))


def alpha_n(V):
    return 0.01 * (V + 55) / (1 - np.exp(-(V + 55) / 10))


def beta_n(V):
    return 0.125 * np.exp(-(V + 65) / 80)


V = np.linspace(-100, 50, 300)
# ----------------------------------------------
```

```python
tau_m = 1 / (alpha_m(V) + beta_m(V))
tau_h = 1 / (alpha_h(V) + beta_h(V))
tau_n = 1 / (alpha_n(V) + beta_n(V))

plt.plot(V, tau_m, label=r"$\tau_m$", color="C0")
plt.plot(V, tau_h, label=r"$\tau_h$", color="C1")
plt.plot(V, tau_n, label=r"$\tau_n$", color="C2")
plt.xlabel("Membrane potential (mV)")
plt.ylabel("Time constant (ms)")
plt.title("HH time constant curves")
plt.legend()
plt.show()
```

## Answers

1. Sodium activation ($\tau_m \sim 0.1$–$0.5$ ms) is roughly 10–20x faster than sodium inactivation ($\tau_h \sim 1$–$8$ ms) and potassium activation ($\tau_n \sim 1$–$6$ ms). This speed difference is essential: the channel must fully open ($m \to 1$) _before_ inactivation shuts it ($h \to 0$). The brief window of time when $m$ is high but $h$ hasn't yet fallen is when the large inward $\text{Na}^+$ current flows.

2. Each time constant peaks near the voltage where $\alpha \approx \beta$, which is where neither rate dominates and the system equilibrates slowest. This corresponds to the steepest part of the respective $m_\infty$, $h_\infty$, or $n_\infty$ curve.
