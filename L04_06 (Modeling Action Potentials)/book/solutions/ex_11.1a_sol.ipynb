# Exercise 11.1a solution

## 1. Equilibrium solution

Setting $\frac{\mathrm{d}m}{\mathrm{d}t} = 0$:

$$0 = \alpha(1 - m) - \beta m = \alpha - \alpha m - \beta m = \alpha - m(\alpha + \beta)$$

Solving for $m$:

$$m_\infty = \frac{\alpha}{\alpha + \beta}$$

This quantity represents the **steady-state open probability** — the fraction of channels that will be open after the system has fully equilibrated at the current conditions.

## 2. Range of $m_\infty$

For $\alpha, \beta > 0$, both the numerator $\alpha$ and denominator $\alpha + \beta$ are positive, and $\alpha < \alpha + \beta$, so:

$$0 < m_\infty < 1$$

This makes biophysical sense because $m_\infty$ is a probability (the fraction of open channels), which must be between 0 and 1.

## 3. Reformulation

Define $\tau_m = \frac{1}{\alpha + \beta}$. Then $m_\infty = \alpha \tau_m$.

Starting from $\frac{\mathrm{d}m}{\mathrm{d}t} = \alpha - m(\alpha + \beta)$:

$$\frac{\mathrm{d}m}{\mathrm{d}t} = \frac{\alpha}{\alpha+\beta} \cdot (\alpha+\beta) - m(\alpha + \beta) = \frac{m_\infty - m}{\tau_m}$$

This is a standard **first-order linear relaxation equation**. It tells us that $m$ exponentially approaches $m_\infty$ with time constant $\tau_m$.
