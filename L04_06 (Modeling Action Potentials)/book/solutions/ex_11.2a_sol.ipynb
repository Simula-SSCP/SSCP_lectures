# Exercise 11.2a solution

## 1. Evaluating the rate function

For $\alpha(V) = e^{(V - V_\alpha)/d_\alpha}$:

- $\alpha(V_\alpha) = e^0 = 1$
- $\alpha(V_\alpha + d_\alpha) = e^1 = e \approx 2.718$
- $\alpha(V_\alpha - d_\alpha) = e^{-1} = 1/e \approx 0.368$

## 2. Monotonicity

- For $d_\alpha > 0$: as $V$ increases, the exponent $(V - V_\alpha)/d_\alpha$ increases, so $\alpha$ is an **increasing** function of $V$. Depolarization speeds up the opening rate.
- For $d_\alpha < 0$: the exponent decreases with $V$, so $\alpha$ is a **decreasing** function of $V$. Depolarization slows down the rate.
