## Phase-plane exploration

### 1. Fixed points with default parameters

With default HH parameters, the two nullclines ($\dot{V}=0$ and $\dot{h}=0$) intersect at **one point**, which corresponds to the resting state of the cell (near $V \approx -65$ mV, $h \approx 0.6$). This single fixed point is **stable** — small perturbations return the system to rest.

### 2. Oscillatory regime

By increasing $\bar{g}_{\mathrm{Na}}$ or shifting $V_{am}$ (the half-activation of $m$) leftward, the stable fixed point can become **unstable**, and the system enters a regime of self-sustained oscillations (repetitive firing). In this regime, the nullclines still intersect at one point, but the fixed point is now an unstable spiral — trajectories spiral outward and settle onto a **limit cycle** (a closed loop in the phase plane representing periodic action potentials).

### 3. Dependence on initial conditions

In the oscillatory regime, all initial conditions eventually converge to the same limit cycle, regardless of where they start. This is a characteristic property of limit cycle oscillators: the cycle is a global attractor (within the physically meaningful region of the phase space).
