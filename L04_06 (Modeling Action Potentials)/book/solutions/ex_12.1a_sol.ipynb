$$I_{\mathrm{app}} = C_m \cdot \frac{\Delta V}{\Delta t} = 0.05 \times 25 = 1.25 \text{ nA}$$

**2.** With $g_{\mathrm{K}} = 0.2$ µS, the potassium current opposes depolarization by pulling $V$ toward $E_K = -80$ mV. Setting $V_{ss} = -40$ mV:

$$I_{\mathrm{app}} = g_K (V_{\mathrm{target}} - E_K) = 0.2 \times 40 = 8 \text{ nA}$$

This is over 6× more current than without the conductance — the potassium leak acts as a current sink that must be overcome.
