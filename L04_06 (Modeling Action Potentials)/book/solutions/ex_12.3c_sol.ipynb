## Bifurcation analysis (optional)

### 1. Bifurcation type

As you slowly increase $\bar{g}_{\mathrm{Na}}$ (or shift $V_{am}$ leftward), the stable fixed point first becomes an unstable spiral, and a stable limit cycle appears simultaneously. This is a **supercritical Hopf bifurcation**: the transition is smooth, with the amplitude of oscillations growing continuously from zero.

Near the bifurcation point, the oscillations are small and sinusoidal. Further from the bifurcation, they become larger and more "action-potential-shaped" (relaxation oscillations with distinct fast and slow phases).

### 2. Bistability

Bistability can occur in models with a **subcritical Hopf bifurcation** or a **saddle-node bifurcation on an invariant circle (SNIC)**. With the simplified two-variable HH model, bistability may be difficult to find with standard parameters, but it can sometimes be achieved by:

- Adding a slow outward current (e.g., an adaptation current)
- Modifying the $h$-nullcline shape by changing the inactivation parameters
- Using a model with a cubic $V$-nullcline that has three intersections with the $h$-nullcline

When bistability exists, the system has two stable states (e.g., resting and oscillating), and the behavior depends on the initial condition. A sufficiently large perturbation can kick the system from one attractor basin to the other.
