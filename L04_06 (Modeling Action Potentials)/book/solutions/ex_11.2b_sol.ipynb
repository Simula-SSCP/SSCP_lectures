## 1. Sharp transition around $V = -50$ mV

To get $m_\infty$ transitioning rapidly near $V = -50$ mV, set $V_\alpha \approx V_\beta \approx -50$ mV with small slope factors (e.g., $d_\alpha = 10$, $d_\beta = -10$). The smaller the magnitude of $d$, the steeper the transition.

## 2. Time constant peak location

The time constant $\tau_m = 1/(\alpha + \beta)$ is maximized when $\alpha + \beta$ is minimized. For exponential rates, this happens near the voltage where $\alpha(V) \approx \beta(V)$. With symmetric slopes ($d_\alpha = -d_\beta$), the crossing point is roughly at $V = (V_\alpha + V_\beta)/2$. Moving $V_\alpha$ or $V_\beta$ shifts the peak accordingly.

The time constant is largest when the two rates are closest in value because neither the opening nor the closing transition dominates — the system "hesitates" the most.

## 3. Decreasing $m_\infty$

To get a decreasing $m_\infty(V)$, make $\alpha$ a decreasing function of $V$ and $\beta$ an increasing function. This requires $d_\alpha < 0$ and $d_\beta > 0$. This represents an **inactivation gate** (like $h$): the gate is open at rest and closes with depolarization.

## 4. Connecting to biology

- **Activation gate $m$:** $\alpha_m$ should increase with depolarization ($d_\alpha > 0$) and $\beta_m$ should decrease ($d_\beta < 0$). Fast time constant.
- **Inactivation gate $h$:** $\alpha_h$ should decrease with depolarization ($d_\alpha < 0$) and $\beta_h$ should increase ($d_\beta > 0$). Slower time constant (larger $|d|$ values).
