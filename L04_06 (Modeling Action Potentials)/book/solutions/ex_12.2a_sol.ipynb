## 1. L'Hôpital's rule

We want $\lim_{x \to 0} \frac{x}{1-e^{-x}}$. Both numerator and denominator go to 0, so we apply L'Hôpital's rule:

$$\lim_{x\to 0}\frac{x}{1-e^{-x}} = \lim_{x\to 0}\frac{1}{e^{-x}} = \frac{1}{1} = 1$$

## 2. Taylor expansion

The Taylor expansion of $e^{-x}$ around $x=0$ is $e^{-x} \approx 1 - x + x^2/2 - \ldots$, so:

$$1 - e^{-x} \approx x - x^2/2$$

Therefore:

$$\frac{x}{1-e^{-x}} \approx \frac{x}{x - x^2/2} = \frac{1}{1 - x/2} \approx 1 + x/2$$

(using $\frac{1}{1-\epsilon} \approx 1 + \epsilon$ for small $\epsilon$).

## 3. Implementation


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp


def vtrap(x):
    """Numerically stable evaluation of x / (1 - exp(-x))."""
    if np.abs(x) < 1e-6:
        return 1.0 + x / 2.0
    else:
        return x / (1.0 - np.exp(-x))


# Test
print(f"vtrap(0.0) = {vtrap(0.0)}")  # Should be ~1.0
print(f"vtrap(1e-8) = {vtrap(1e-8)}")  # Should be ~1.0
print(f"vtrap(1.0) = {vtrap(1.0):.6f}")  # Should be ~1.5820
```
