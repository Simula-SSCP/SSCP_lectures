# Exercise 12.2b solution


The problematic rate functions are $\alpha_m$ and $\alpha_n$, which both have the form $\frac{a(V + b)}{1 - e^{-(V+b)/c}}$. We can rewrite them using vtrap by setting $u = (V + b)/c$. Specifically, by setting $u = (V+40)/10$, we can rewrite:

$$\alpha_m(V) = \frac{0.1(V+40)}{1 - e^{-(V+40)/10}} = \frac{0.1 \cdot 10u}{1 - e^{-u}} = \frac{u}{1 - e^{-u}} = \text{vtrap}(u)$$

So: $\alpha_m(V) = \text{vtrap}\!\left(\frac{V+40}{10}\right)$

Similarly for $\alpha_n$, with $u = (V+55)/10$:

$$\alpha_n(V) = \frac{0.01(V+55)}{1 - e^{-(V+55)/10}} = \frac{0.01 \cdot 10u}{1 - e^{-u}} = \frac{0.1u}{1 - e^{-u}} = 0.1 \cdot \text{vtrap}(u)$$

So: $\alpha_n(V) = 0.1 \cdot \text{vtrap}\!\left(\frac{V+55}{10}\right)$


```python
# --- Setup code from previous sub-exercises ---
import numpy as np


def vtrap(x):
    """Numerically stable evaluation of x / (1 - exp(-x))."""
    if np.abs(x) < 1e-6:
        return 1.0 + x / 2.0
    else:
        return x / (1.0 - np.exp(-x))


# ----------------------------------------------
```

```python
def alpha_m(V):
    return vtrap((V + 40) / 10)


def alpha_n(V):
    return 0.1 * vtrap((V + 55) / 10)


# Verify: these should produce sensible values at the singular points
print(f"alpha_m(-40) = {alpha_m(-40):.4f}")  # Singular point, should be ~1.0
print(f"alpha_m(-50) = {alpha_m(-50):.4f}")  # Normal point
print(f"alpha_n(-55) = {alpha_n(-55):.4f}")  # Singular point, should be ~0.1
print(f"alpha_n(-65) = {alpha_n(-65):.4f}")  # Normal point
```

Both produce identical results to the original formulas at all non-singular voltages, and return the correct limit at the singular points ($V = -40$ mV for $\alpha_m$ and $V = -55$ mV for $\alpha_n$).
