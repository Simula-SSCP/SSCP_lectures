# Exercise 13.1c solution

## Ventricular vs. atrial cells


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

import GBV_RHS as gb
import GBV_D as init


def run_grandi_bers(conductance_multipliers):
    y0 = init.get_initial_conditions()
    params = gb.get_default_parameters()
    if "g_CaL" in conductance_multipliers:
        params["g_CaL"] *= conductance_multipliers["g_CaL"]
    if "g_Kr" in conductance_multipliers:
        params["g_Kr"] *= conductance_multipliers["g_Kr"]
    if "g_to" in conductance_multipliers:
        params["g_to"] *= conductance_multipliers["g_to"]
    t_span = (0, 600)
    sol = solve_ivp(gb.rhs, t_span, y0, args=(params,), max_step=1.0)
    return sol.t, sol.y[0]


# Run baseline, Long QT, and atrial-like simulations
t_base, V_base = run_grandi_bers({"g_CaL": 1.0, "g_Kr": 1.0, "g_to": 1.0})
t_atrial, V_atrial = run_grandi_bers({"g_CaL": 0.4, "g_Kr": 1.0, "g_to": 3.0})

# Plot all three
plt.figure(figsize=(8, 5))
plt.plot(t_base, V_base, "k-", linewidth=2, label="Ventricular (baseline)")
plt.plot(t_atrial, V_atrial, "b-", linewidth=2, label="Atrial-like (↓CaL, ↑Ito)")
plt.xlabel("Time (ms)")
plt.ylabel("Membrane Potential (mV)")
plt.title("Ventricular vs. atrial-like action potentials")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### Observations

The "atrial-like" AP is dramatically shorter and has a very different morphology:

- **Prominent notch:** The enlarged $I_{\mathrm{to}}$ creates a deep early repolarization notch immediately after the upstroke.
- **Short/absent plateau:** The reduced $I_{\mathrm{CaL}}$ cannot sustain the prolonged plateau that characterizes ventricular cells. Without sufficient inward calcium current to oppose the outward potassium currents, the cell repolarizes quickly.
- **Shorter APD:** The overall action potential duration is much shorter ($\sim 100$–$150$ ms vs. $\sim 300$ ms for ventricular cells).

This exercise demonstrates a key principle: the shape and duration of the action potential is determined by the _balance_ of inward and outward currents, which differs between cell types. Different cells in the heart (atrial, ventricular, Purkinje, nodal) express different combinations of ion channels, producing the diverse AP shapes needed for coordinated cardiac function.

Importantly, this also means that drugs targeting specific channels can have different effects in different parts of the heart—a critical consideration in cardiac pharmacology.
