## Excitability and threshold

### 1. Small vs. large perturbations

A small depolarizing perturbation (e.g., from $-65$ to $-60$ mV) causes a small, passive return to rest. The trajectory in the phase plane curves smoothly back to the fixed point without completing a large excursion.

A larger perturbation (e.g., to $-55$ mV or beyond) triggers a full action potential: the trajectory swings far to the right (depolarization), loops around at high $V$ and low $h$, then returns to rest via the afterhyperpolarization. This is the hallmark of an **excitable system** — the response is all-or-nothing.

### 2. Threshold

The threshold is approximately $-55$ to $-50$ mV with default parameters, though it is not a single sharp voltage. In the phase plane, the threshold corresponds to the **separatrix** — an invisible boundary curve. Perturbations that stay on the "near" side of the separatrix return directly to rest; those that cross it trigger the full excursion. The separatrix approximately follows the middle branch of the $V$-nullcline (if it is N-shaped).

### 3. The four AP phases in the phase plane

The trajectory of a single action potential traces a characteristic loop:

1. **Upstroke:** Rapid rightward motion (increasing $V$) at nearly constant $h$ — the $m$ gate opens much faster than $h$ changes.
2. **Peak and early repolarization:** The trajectory curves downward (decreasing $h$) as inactivation kicks in. $V$ begins to decrease.
3. **Repolarization:** Leftward motion (decreasing $V$) with $h$ still low — the $K^+$ current drives repolarization.
4. **Recovery (afterhyperpolarization):** At low $V$, the trajectory curves upward (increasing $h$) as the inactivation gate recovers. The system slowly returns to the resting fixed point.
