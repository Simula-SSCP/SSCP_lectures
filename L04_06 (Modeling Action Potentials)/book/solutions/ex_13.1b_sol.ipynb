## Simulating Long QT syndrome


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp

import GBV_RHS as gb
import GBV_D as init


def run_grandi_bers(conductance_multipliers):
    y0 = init.get_initial_conditions()
    params = gb.get_default_parameters()
    if "g_CaL" in conductance_multipliers:
        params["g_CaL"] *= conductance_multipliers["g_CaL"]
    if "g_Kr" in conductance_multipliers:
        params["g_Kr"] *= conductance_multipliers["g_Kr"]
    if "g_to" in conductance_multipliers:
        params["g_to"] *= conductance_multipliers["g_to"]
    t_span = (0, 600)
    sol = solve_ivp(gb.rhs, t_span, y0, args=(params,), max_step=1.0)
    return sol.t, sol.y[0]


# Run baseline and Long QT simulations
t_base, V_base = run_grandi_bers({"g_CaL": 1.0, "g_Kr": 1.0, "g_to": 1.0})
t_lqt, V_lqt = run_grandi_bers({"g_CaL": 1.0, "g_Kr": 0.3, "g_to": 1.0})

# Plot
plt.figure(figsize=(8, 5))
plt.plot(t_base, V_base, "k--", linewidth=1.5, label="Baseline")
plt.plot(t_lqt, V_lqt, "r-", linewidth=2, label="Long QT (70% IKr block)")
plt.xlabel("Time (ms)")
plt.ylabel("Membrane Potential (mV)")
plt.title("Drug-induced Long QT syndrome")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### Observations

Reducing $I_{\mathrm{Kr}}$ by 70% dramatically prolongs the action potential. The plateau phase extends significantly because there is not enough outward potassium current to overcome the inward calcium current and drive repolarization.

This is exactly what happens in drug-induced Long QT syndrome:
- The prolonged AP means the heart takes longer to "reset" between beats
- On an ECG, this manifests as a prolonged QT interval
- In severe cases, the prolonged plateau can trigger early afterdepolarizations (EADs) — spontaneous re-excitations during the plateau that can degenerate into the lethal arrhythmia Torsades de Pointes

This is why every new drug candidate is screened for hERG ($I_{\mathrm{Kr}}$) channel block during development.
