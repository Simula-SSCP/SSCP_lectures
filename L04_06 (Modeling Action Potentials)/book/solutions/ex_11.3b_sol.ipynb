# Exercise 11.3b solution


```python
# --- Setup code from previous sub-exercises ---
import numpy as np
import matplotlib.pyplot as plt


def alpha_m(V):
    return 0.1 * (V + 40) / (1 - np.exp(-(V + 40) / 10))


def beta_m(V):
    return 4.0 * np.exp(-(V + 65) / 18)


def alpha_h(V):
    return 0.07 * np.exp(-(V + 65) / 20)


def beta_h(V):
    return 1.0 / (1 + np.exp(-(V + 35) / 10))


def alpha_n(V):
    return 0.01 * (V + 55) / (1 - np.exp(-(V + 55) / 10))


def beta_n(V):
    return 0.125 * np.exp(-(V + 65) / 80)


# ----------------------------------------------
```

```python
V = np.linspace(-100, 50, 300)

m_inf = alpha_m(V) / (alpha_m(V) + beta_m(V))
h_inf = alpha_h(V) / (alpha_h(V) + beta_h(V))
n_inf = alpha_n(V) / (alpha_n(V) + beta_n(V))

plt.plot(V, m_inf, label=r"$m_\infty$", color="C0")
plt.plot(V, h_inf, label=r"$h_\infty$", color="C1")
plt.plot(V, n_inf, label=r"$n_\infty$", color="C2")
plt.xlabel("Membrane potential (mV)")
plt.ylabel("Steady-state value")
plt.title("HH steady-state gating curves")
plt.legend()
plt.show()
```

## Answers

1. At $V = -65$ mV: $m_\infty \approx 0.05$, $h_\infty \approx 0.60$, $n_\infty \approx 0.32$. At rest, almost all $m$ gates are closed, most $h$ gates are available (not inactivated), and about a third of $n$ gates are open.

2. $m_\infty = 0.5$ occurs near $V \approx -40$ mV. For $n_\infty$, it occurs near $V \approx -55$ mV. The potassium activation is shifted leftward.

3. The sodium window current region is roughly $-60$ to $-40$ mV, where both $m_\infty > 0$ and $h_\infty > 0$ simultaneously.
