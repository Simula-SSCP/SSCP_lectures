# Exercise 12.1: Firing the Hodgkin-Huxley action potential

We are now going to solve the full Hodgkin-Huxley model from scratch. The system consists of four coupled ordinary differential equations: one for the membrane voltage $V$, and one for each of the three gating variables ($m$, $h$, and $n$).

$$
C_{\mathrm{m}}\frac{\mathrm{d}V}{\mathrm{d}t} = -\big(I_{\mathrm{Na}} + I_{\mathrm{K}} + I_{\mathrm{L}} + I_{\mathrm{app}}\big)
$$

$$
\frac{\mathrm{d}m}{\mathrm{d}t} = \alpha_m (1-m) - \beta_m m
$$

$$
\frac{\mathrm{d}h}{\mathrm{d}t} = \alpha_h (1-h) - \beta_h h
$$

$$
\frac{\mathrm{d}n}{\mathrm{d}t} = \alpha_n (1-n) - \beta_n n
$$

The mathematical formulations for all six $\alpha$ and $\beta$ transition rates are provided in the code cell below.
