# Exercise 12.3: Phase-plane analysis of excitability

In Exercise 12.1, we built up to the full HH model by adding conductances and gates one at a time. In this exercise, we will take a different perspective and analyze a simplified two-variable excitable model using **phase-plane analysis**.

Phase-plane analysis is a powerful mathematical tool for understanding the qualitative behavior of dynamical systems without solving the equations numerically. By plotting the _nullclines_ (the curves where $\dot{V} = 0$ and $\dot{h} = 0$), we can identify fixed points, predict stability, and understand when a system can oscillate.


## 12.3a: The reduced model

We consider the following reduced two-variable model. Because sodium activation ($m$) is much faster than all other gates, we replace it with its steady-state approximation $m \approx m_\infty(V)$. This eliminates one ODE and leaves us with a system of two:

$$C_m \frac{\mathrm{d}V}{\mathrm{d}t} = -\bar{g}_{\mathrm{Na}} \, m_\infty(V) \, h \, (V - E_{\mathrm{Na}}) - \bar{g}_{\mathrm{K}} (V - E_{\mathrm{K}})$$

$$\frac{\mathrm{d}h}{\mathrm{d}t} = \alpha_h(V)(1-h) - \beta_h(V) h$$

This is a simplified model that captures the essential excitability of the membrane using just $V$ and $h$.
